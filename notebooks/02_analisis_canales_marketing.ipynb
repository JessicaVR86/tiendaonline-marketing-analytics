```python
df = spark.table("workspace.default.sesiones_web")
df.createOrReplaceTempView("sesiones_web")
print(f"Total de sesiones: {df.count()}")   
print(f"Total de conversiones: {df.filter('conversion == true').count()}")  
```

```python
%sql
SELECT
    canal,
    COUNT(*) AS total_sesiones,
    SUM(CASE WHEN conversion THEN 1 ELSE 0 END) AS total_conversiones,
    ROUND(AVG(CASE WHEN conversion THEN 1.0 ELSE 0.0 END) * 100, 2) AS tasa_conversion_pct,
    ROUND(SUM(valor_compra), 2) AS ingresos_totales
FROM sesiones_web
GROUP BY canal
ORDER BY total_sesiones DESC
```

```python
%sql
SELECT
    dispositivo,
    COUNT(*) AS total_sesiones,
    ROUND(AVG(CASE WHEN conversion THEN 1.0 ELSE 0.0 END) * 100, 2) AS tasa_conversion_pct,
    ROUND(AVG(duracion_segundos), 1) AS duracion_promedio_seg
FROM sesiones_web
GROUP BY dispositivo
ORDER BY total_sesiones DESC
```

```python
%sql
SELECT
    canal,
    campana,
    COUNT(*) AS total_sesiones,
    SUM(CASE WHEN conversion THEN 1 ELSE 0 END) AS conversiones,
    ROUND(AVG(CASE WHEN conversion THEN 1.0 ELSE 0.0 END) * 100, 2) AS tasa_conversion_pct
FROM sesiones_web
GROUP BY canal, campana
HAVING COUNT(*) > 50
ORDER BY tasa_conversion_pct DESC
```
